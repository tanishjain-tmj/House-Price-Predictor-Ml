#Project House Price Predictor

```python
import pandas as pd
import numpy as np
```

```python
house = pd.read_csv('/content/Bengaluru_House_Data (1).csv')
```

```python
house.head()
```

```python
house.shape
```

```python
house.info()
```

```python
# creating backup copy
backup = house.copy()
```

```python
# checking missing values
house.isnull().sum()
```

```python
# removing society because it has many missing values
house.drop('society', axis=1, inplace=True)
```

```python
# removing remaining missing values
house = house.dropna()
```

```python
house = house.reset_index(drop=True)
```

```python
house.isnull().sum()
```

```python
#data type cleaning
house.dtypes
```

```python
# extracting BHK from size
house['BHK'] = house['size'].str.extract(r'(\d+)').astype(float)
```

```python
house[['size', 'BHK']].head()
```

```python
house.drop('size', axis=1, inplace=True)
```

```python
#clean total_sqft
def convert_sqft(x):
    try:
        if '-' in str(x):
            a, b = str(x).split('-')
            return (float(a) + float(b)) / 2
        else:
            return float(x)
    except:
        return np.nan
```

```python
house['total_sqft'] = house['total_sqft'].apply(convert_sqft)
```

```python
house.isnull().sum()
```

```python
house = house.dropna()
```

```python
house = house.reset_index(drop=True)
```

```python
#target column
house.rename(columns={'price': 'Price'}, inplace=True)
```

```python
house.head()
```

```python
house.info()
```

```python
#Outlier removal
house['Price'].describe()
```

```python
Q1 = house['Price'].quantile(0.25)
Q3 = house['Price'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

house = house[
    (house['Price'] >= lower) &
    (house['Price'] <= upper)
]
```

```python
house = house.reset_index(drop=True)
```

```python
house.head()
```

```python
house.info()
```

```python
house.describe(include='all')
```

```python
house.to_csv(
    'Cleaned_House_data.csv',
    index=False
)
```

```python
# extracting training data
X = house.drop('Price', axis=1)
Y = house['Price']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
categorical_cols = X.select_dtypes(
    include='object'
).columns

categorical_cols
```

```python
numerical_cols = X.select_dtypes(
    include=np.number
).columns

numerical_cols
```

```python
#model building
#onehotencoder
from sklearn.preprocessing import OneHotEncoder

one = OneHotEncoder(
    handle_unknown='ignore'
)
```

```python
#column transformer
from sklearn.compose import make_column_transformer

column_trans = make_column_transformer(
    (one, categorical_cols),
    remainder='passthrough'
)
```

```python
#linear regression
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
```

```python
#pipeline
from sklearn.pipeline import make_pipeline

pipe = make_pipeline(
    column_trans,
    lr
)
```

```python
#fit the model
pipe.fit(X_train, Y_train)
```

```python
#predction
Y_pred = pipe.predict(X_test)
```

```python
#r2 score
from sklearn.metrics import r2_score

r2_score(Y_test, Y_pred)
```

```python
#find best random state
scores = []

for i in range(1000):

    X_train, X_test, Y_train, Y_test = train_test_split(
        X,
        Y,
        test_size=0.2,
        random_state=i
    )

    lr = LinearRegression()

    pipe = make_pipeline(
        column_trans,
        lr
    )

    pipe.fit(X_train, Y_train)

    Y_pred = pipe.predict(X_test)

    scores.append(
        r2_score(Y_test, Y_pred)
    )
```

```python
np.argmax(scores)
```

```python
scores[np.argmax(scores)]
```

```python
best_random_state = np.argmax(scores)

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=best_random_state
)
```

```python
lr = LinearRegression()

pipe = make_pipeline(
    column_trans,
    lr
)

pipe.fit(X_train, Y_train)
```

```python
from sklearn.metrics import r2_score
Y_pred = pipe.predict(X_test)
print("R2 Score:", r2_score(Y_test, Y_pred))
```
